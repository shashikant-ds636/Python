```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
data={
    "python":[ "yes","yes","no","yes","no","yes","no","no","yes","yes","yes","no","yes","no","no","yes","yes","no","no","yes" ],
    "machine learning":[ "yes","yes","no","yes","no","no","no","no","yes","yes","yes","no","yes","no","no","yes","yes","no","no","yes" ],
    "experince":[2,1,0,3,2, 0,4,1,2,3, 0,5,2,1,4, 3,2,0,5,1],
    "projects":[3,2,1,4,5, 0,4,1,2,5, 0,5,3,1,4, 3,2,0,5,1],
    "shortlisted":[ "yes","yes","no","yes","no","yes","no","no","yes","yes","yes","no","yes","no","no","yes","yes","no","no","yes" ] }
```

```python
df=pd.DataFrame(data)
df
```

appling encoders on data

```python
encode=LabelEncoder()
df["python"]=encode.fit_transform(df["python"])
df["machine learning"]=encode.fit_transform(df["machine learning"])
df["shortlisted"]=encode.fit_transform(df["shortlisted"])
df
```

```python
X=df.drop("shortlisted",axis=1)
y=df["shortlisted"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=9)
```

```python
model = GaussianNB()
model.fit(X_train, y_train)
```

```python
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print(f"Name of Model: Gaussian NB and the Accuracy score: {accuracy_score(y_test, predictions)}")
print()
print(classification_report(y_test, predictions))
train = model.score(X_train, y_train)
test = model.score(X_test, y_test)
print(f"Train Score: {train} and Test Score: {test}")
```

```python
def mymodel():
  name = input("Enter your name: ")
  # Changed inputs to expect 'yes'/'no' strings for categorical features
  python_val = input("Enter your value for python (yes/no): ")
  machine_learning_val = input("Enter your value for machine learning (yes/no): ")
  experince = int(input("Enter your value for experience: "))
  projects = int(input("Enter your value for projects: "))

  # Create a DataFrame from the new observation
  new_data = {
      "python": [python_val],
      "machine learning": [machine_learning_val],
      "experince": [experince],
      "projects": [projects]
  }
  df1 = pd.DataFrame(new_data)

  # Apply LabelEncoder to the categorical columns of the new DataFrame
  # It's important to use the same encoding logic as the training data.
  # Since LabelEncoder sorts categories, 'no' will be 0 and 'yes' will be 1.
  # We create a temporary encoder for this, assuming 'no' and 'yes' are the only categories.
  temp_encoder = LabelEncoder()
  # Fit to the possible categories ['no', 'yes'] to ensure consistent mapping
  temp_encoder.fit(['no', 'yes'])
  df1["python"] = temp_encoder.transform(df1["python"])
  df1["machine learning"] = temp_encoder.transform(df1["machine learning"])

  # Predict using the processed DataFrame df1
  predictions = model.predict(df1)

  if predictions[0] == 1:
    print(f"Hello {name}")
    print("You are shortlisted!")
  else:
    print(f"Hello {name}")
    print("You are not shortlisted.")
```
